```python
import pandas as pd
import numpy as np

# Load dataset
matches = pd.read_csv('D:/IPL-PREDICTION-PROJECT/data/raw/ipl.csv')

# Rename column
matches = matches.rename(columns={
    'match_date': 'date'
})

# Check missing values
print("Missing Values:\n")
print(matches.isnull().sum())

# Standardize team names
team_replacements = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Rising Pune Supergiant': 'Rising Pune Supergiants'
}

matches.replace(team_replacements, inplace=True)

# Keep important columns
matches = matches[[
    'season',
    'team1',
    'team2',
    'toss_winner',
    'toss_decision',
    'venue',
    'winner',
    'result',
    'result_margin',
    'target_runs'
]]

# Display first rows
print("\nCleaned Dataset Preview:\n")
print(matches.head())

# Save cleaned dataset
matches.to_csv('D:/IPL-PREDICTION-PROJECT/data/processed/cleaned_matches.csv', index=False)

print("\n✅ Cleaned dataset saved successfully.")
```
