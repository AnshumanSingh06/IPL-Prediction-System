```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
matches = pd.read_csv('D:/IPL-PREDICTION-PROJECT/data/processed/cleaned_matches.csv')

# Set style
sns.set_style('darkgrid')
```

```python
# 1. Most Successful IPL Teams
# ==============================

winner_counts = matches['winner'].value_counts()

plt.figure(figsize=(12,6))
sns.barplot(
    x=winner_counts.index,
    y=winner_counts.values
)

plt.xticks(rotation=90)
plt.title('Most Successful IPL Teams')
plt.xlabel('Teams')
plt.ylabel('Number of Wins')

plt.show()
```

```python
# 2. Toss Decision Analysis
# ==============================

toss_counts = matches['toss_decision'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    toss_counts.values,
    labels=toss_counts.index,
    autopct='%1.1f%%'
)

plt.title('Toss Decision Distribution')
plt.show()
```

```python
# 3. Top IPL Venues
# ==============================

top_venues = matches['venue'].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot(
    x=top_venues.values,
    y=top_venues.index
)

plt.title('Top 10 IPL Venues')
plt.xlabel('Matches Hosted')
plt.ylabel('Venue')

plt.show()
```

```python
# 4. Toss Winner vs Match Winner
# ==============================

toss_match_win = (
    matches['toss_winner'] == matches['winner']
).mean() * 100

print(f'Teams winning toss also won the match in {toss_match_win:.2f}% matches')


```

```python
# 5. Matches Played Per Season
# ==============================

season_counts = matches['season'].value_counts().sort_index()

plt.figure(figsize=(12,6))
sns.lineplot(
    x=season_counts.index,
    y=season_counts.values,
    marker='o'
)

plt.xticks(rotation=45)
plt.title('IPL Matches Per Season')
plt.xlabel('Season')
plt.ylabel('Number of Matches')

plt.show()
```
